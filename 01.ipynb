```python
# Data Visualization – 3 Important MCQs with Answers

# Q1. An EdTech company wants to analyze students' marks to find the range in which most students scored. Which visualization is best?
# A. Pie Chart
# B. Histogram
# C. Countplot
# D. Lineplot
# Correct Answer: B. Histogram
# Reason: A histogram shows the distribution of marks across different ranges or intervals.


# Q2. Airtel wants to find the count of customers across different payment modes used by customers. Which visualization is best?
# A. Pie Chart
# B. Countplot
# C. Lineplot
# D. Boxplot
# Correct Answer: B. Countplot
# Reason: A countplot displays the number of customers in each payment category.


# Q3. In the state of Haryana, we want to find the proportion of people who smoke. Which visualization is best?
# A. Pie Chart
# B. Bar Chart
# C. Countplot
# D. Boxplot
# Correct Answer: A. Pie Chart
# Reason: A pie chart shows the proportion or percentage of smokers and non-smokers.

# Quick Revision Table
# Visualization	Main Purpose
# Histogram	Distribution across ranges
# Countplot	Count of categories
# Pie Chart	Proportion / Percentage
# Bar Chart	Comparison between categories
# Lineplot	Trends over time
# Boxplot	Distribution and outliers

# Exam Tip:
# - Range / Distribution → Histogram
# - Count / Frequency → Countplot
# - Proportion / Percentage → Pie Chart
# - Comparison → Bar Chart
# - Trend → Lineplot
# - Outliers → Boxplot
```

```python
top_5_pub, top_5_plat, top_5_genre
```

```python
# top 5 data 
top_5_data = data.loc[(data['Publisher'].isin(top_5_pub))& (data['Platform'].isin(top_5_plat)) & (data['Genre'].isin(top_5_genre))]
top_5_data
```

```python
# Draw Dodge Plot
sns.countplot(data = top_5_data, x = "Publisher", hue = "Platform")  
```

```python
# Stakced Plot
df_stacked_plt = pd.crosstab(index = top_5_data['Publisher'], columns = top_5_data["Platform"])
```

```python
df_stacked_plt
```

```python
df_stacked_plt.plot(kind = "bar", stacked = True)
plt.show()
```
